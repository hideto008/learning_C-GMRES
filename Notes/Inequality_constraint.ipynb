# 不等式制約の考慮

## KKT 条件の再考

KKT条件は不等式制約 $g \le 0$ のもと、関数 $f$ の最小値を探す問題である。

その必要条件として以下の四つの式を満たす必要がある。

$$
\begin{array}{l}
\nabla f + \lambda \nabla g = 0 \\
g \le 0 \\
\lambda \ge 0 \\
\lambda g = 0 \\
\end{array}
$$

四つ目の式は相補性条件 $\lambda g =0 $ であり、これは、制約条件の境界上にであれば $g=0$ となり、境界上でなければ相補性条件を満たすため$\lambda = 0$ となる。

## PMPにおける最適制御問題の再考

PMPは $J$ の下記の制約付き最小化である。

$$
\begin{array}{c}
\min\limits_{u} \ J = \Phi(x(T)) + \int_t^T L(x,)d \tau \\
\text{sujbect to } \quad \dot{x} = f(x,u)
\end{array}
$$
