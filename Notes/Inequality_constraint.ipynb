# 不等式制約の考慮 (ダミー変数)

## KKT 条件の再考

KKT条件は不等式制約 $g(x) \le 0$ のもと、関数 $f(x)$ の最小値を探す問題である。

その必要条件として以下の四つの式を満たす必要がある。[KKT](KKT.ipynb)のNoteでは随伴変数を$\lambda$で記載していたが、KKT条件をPMPへ組み込むことを考慮し、$\mu$としている。

||式|名前|
|:---|:---|:---|
|(1)|$\nabla f + \mu \nabla g = 0$| Stationarity (停留条件)|
|(2)|$g \le 0$|Primal feasibility (主問題の実行可能性)|
|(3)|$\mu \ge  0$|Dual feasibility (双対実行可能性)|
|(4)|$\mu g = 0$| Complementary slackness(相補性条件)|

重要なのは四つ目の相補性条件 $\mu g =0 $ であり、これは、制約条件の境界上であれば $g=0$ となり、境界上でなければ相補性条件を満たすため$\mu = 0$ となるものである。

もう少し説明を追加する。 $x$の状態により、$g = -2$ のように制約の内側にあるときは $x$ は自由に動くことができる。これを制約が非アクティブであるという。ここで $\mu g = 0$ の相補性条件より、$\mu=0$ でなければならない。
また、$g=0$ と制約の境界上にいる場合、制約の外側、つまり $g > 0$ へ動こうとすると、制約の内側へ戻そうとする力$\mu \ge 0$ により$g=0$ へと押し留まるような意味になる。これをアクティブであるという。このとき相補性条件 とDual feasibility より、$\mu \ge 0$ となる。

## PMPにおける最適制御問題の再考

不等式制約 $g(x,u) \le 0$ を組み込んだ最適制御問題は下記のようになる。

$$
\begin{array}{c}
\min\limits_{u} \ J = \Phi(x(T)) + \int_0^T L(x,u) dt \\
\text{subject to } \quad \begin{split} &\dot{x} = f(x,u) \\ &g(x,u) \le 0 \end{split} 
\end{array}
$$

不等式制約 $g(x,u)$は下記のように複数の条件であり、それに伴い$\mu$もベクトルとなる。

$$
g(x,u) = \begin{bmatrix}
g_1(x,u) \\ \vdots \\ g_m(x,u) 
\end{bmatrix} \in \mathbb{R}^{m} , \quad 
\mu = \begin{bmatrix}
\mu_1 \\ \vdots \\ \mu_m
\end{bmatrix} \in \mathbb{R}^m
$$

[PMP](PMP.ipynb)のときと同様に、拡大評価関数$\bar{J}$を構成すると以下のようになる。

$$
\bar{J} = \Phi(x(T)) + \int_0^T \left[ L(x,u) + \lambda^T(f(x,u) - \dot{x}) + \mu^T g(x,u)\right] dt
$$

$\bar{J}$の変分を計算する。

まず、状態 $x$, 制御入力 $u$, ラグランジュ未定乗数 $\lambda$ , $\mu$ をそれぞれ以下のように微小変化させる。

$$
\begin{aligned}
x &\rightarrow x + \varepsilon \delta x \\
\dot{x} &\rightarrow \dot{x} + \varepsilon \delta \dot{x} \\
x(T) &\rightarrow x(T) + \varepsilon \delta x(T) \\
u &\rightarrow u + \varepsilon \delta u \\
\lambda &\rightarrow \lambda + \varepsilon \delta \lambda \\
\mu &\rightarrow \mu + \varepsilon \delta \mu \\
\end{aligned}
$$

したがって、微小変化を代入した拡大評価関数 $\bar{J}$ は以下となる。

$$
\begin{aligned}
\bar{J} & \left[x + \varepsilon \delta x, u + \varepsilon \delta u, \lambda + \varepsilon \delta \lambda,  \mu + \varepsilon \delta \mu  \right] \\
&= \Phi(x(T)+ \varepsilon \delta x(T))  \\ 
 &+  \int_0^T \bigg[ L(x+ \varepsilon \delta x ,u+ \varepsilon \delta u) \\
 &\quad + (\lambda+ \varepsilon \delta \lambda )^T \bigg(f(x+ \varepsilon \delta x ,u+ \varepsilon \delta u) - \dot{x}- \varepsilon \delta \dot{x} \bigg)  \\
 &\qquad + (\mu+ \varepsilon \delta \mu )^T g(x+ \varepsilon \delta x ,u+ \varepsilon \delta u )\bigg] dt
\end{aligned}
$$


変分$\delta$を計算するとは、以下のように、$\varepsilon$ で微分した後、$\varepsilon=0$ と代入する計算であった。

$$
\delta = \left. \frac{d}{d \varepsilon} \right|_{\varepsilon=0}
$$

微小変化を代入した$\bar{J}$の変分を行う。

左辺は以下となる。

$$
\delta \bar{J} = \left. \frac{d}{d \varepsilon} \bar{J}[x + \varepsilon \delta x, u + \varepsilon \delta u, \lambda + \varepsilon \delta \lambda,  \mu + \varepsilon \delta \mu ] \right|_{\varepsilon=0}
$$

右辺は以下となる。

$$
\begin{aligned}
\delta \bar{J} &= \dfrac{d}{d \varepsilon} \Bigg[ \Phi(x(T)+ \varepsilon \delta x(T)) + \int_0^T \bigg[ L(x+ \varepsilon \delta x ,u+ \varepsilon \delta u) \\
 & \quad + (\lambda+ \varepsilon \delta \lambda )^T \bigg(f(x+ \varepsilon \delta x ,u+ \varepsilon \delta u) - \dot{x}- \varepsilon \delta \dot{x} \bigg)  \\
 & \qquad + (\mu+ \varepsilon \delta \mu )^T g(x+ \varepsilon \delta x ,u+ \varepsilon \delta u) \left. \bigg] dt \Bigg] \right|_{\varepsilon=0}
\end{aligned} 
$$

上式は等式制約でのPMPを求める際に、 $\mu^T g(x,u)$の項以外をすでに求めている。よって、以下の変分を考える。

$$
\left. \frac{d}{d \varepsilon} \left[ (\mu+ \varepsilon \delta \mu )^T g(x+ \varepsilon \delta x ,u+ \varepsilon \delta u \right]  \right|_{\varepsilon=0}
$$

等式制約に対するラグランジュ乗数$\lambda$は自由に動かすこと(変分)ができるが、不等式制約に対する $\mu$ は $\mu \ge 0$ という Dual feasability の制約があり、自由に変分することが出来ない。よって、KKT条件を考慮する必要がある。

## 不等式制約を組み込むときの注意

上記では不等式制約 $g(x,u) \le 0$ を 拡張評価関数 $\bar{J}$ へ $\mu^T g(x,u)$ として組み込んだ。

これは $g(x,u)$ がスカラーではなくベクトルであり、不等式制約が複数設定してあるということになる。

$x = (x_1, x_2)$ と状態が複数ある場合、不等式制約は通常 $g_i(x,u) = x_1^2 + x_2^2 -u^2 - 1 \le 0$ のようなスカラーで評価する形をとる。

よって、KKT条件で上げた4式はスカラーの表現である。これを複数の不等式制約とする場合、4式がそれぞれの制約で組み込まれる。

$$
\begin{split}
\begin{array}{l}
\nabla f(x,u) +J_g(x,u)^T \mu = 0 \\
g_i (x,u) \le 0 \\
\mu_i g_i(x,u) = 0 \\
\mu_i \ge 0
\end{array}
\end{split} \quad, i = 1, \cdots, m
$$

式中の$J_g(x,u)^T$は $\nabla g_i$ で構成されたヤコビアンであり、

$$
J_g(x,u) =
\begin{bmatrix}
\nabla g_1 ^T \\
\vdots \\
\nabla g_m ^T
\end{bmatrix} =
\begin{bmatrix}
\dfrac{\partial g_1}{\partial x} & \dfrac{\partial g_1}{\partial u} \\
\vdots & \vdots\\
\dfrac{\partial g_m}{\partial x} & \dfrac{\partial g_m}{\partial u} \\
\end{bmatrix}
$$

$\mu$は$g_i(x,u)$のそれぞれと積算されるスカラー$\mu_i$で構成されたベクトルである。

$$
\mu = \begin{bmatrix} \mu_1 & \cdots & \mu_m\end{bmatrix}^T
$$

よって $J_g(x,u)^T {\mu}$ は以下のようになる。

$$
J_g(x,u)^T {\mu}=
\begin{bmatrix}
\dfrac{\partial g_1}{\partial x} & \cdots & \dfrac{\partial g_m}{\partial x} \\
\dfrac{\partial g_1}{\partial u} & \cdots & \dfrac{\partial g_m}{\partial u} \\
\end{bmatrix}
\begin{bmatrix}
\mu_1 \\ \vdots \\ \mu_m 
\end{bmatrix} = 
\begin{bmatrix}
\mu_1 \dfrac{\partial g_1}{\partial x} + \cdots + \mu_m \dfrac{\partial g_m}{\partial x} \\
\vdots \\
\mu_1 \dfrac{\partial g_1}{\partial u} + \cdots + \mu_m \dfrac{\partial g_m}{\partial u}
\end{bmatrix} =
\mu_1 \begin{bmatrix}
\dfrac{\partial g_1}{\partial x} \\ \dfrac{\partial g_1}{\partial u}
\end{bmatrix} + \cdots +
\mu_m \begin{bmatrix}
\dfrac{\partial g_m}{\partial x} \\ \dfrac{\partial g_m}{\partial u}
\end{bmatrix} 
$$


### 不等式制約に関する変分

$$
\delta ( \mu^T g) = \left. \frac{d}{d \varepsilon} \left[ (\mu+ \varepsilon \delta \mu )^T g(x+ \varepsilon \delta x ,u+ \varepsilon \delta u \right]  \right|_{\varepsilon=0}
$$

上式の計算を積の微分とチェインルールにより計算する。$F(\varepsilon)$ を以下のようにする。$g(x,u), \mu$はそれぞれベクトルであることに注意する。

$$
F(\varepsilon) = (\mu+ \varepsilon \delta \mu )^T g(x+ \varepsilon \delta x ,u+ \varepsilon \delta u)
$$

$$
\frac{d F(\varepsilon)}{d \varepsilon} = \delta \mu^T g(x+ \varepsilon \delta x ,u+ \varepsilon \delta u) + (\mu+ \varepsilon \delta \mu )^T \frac{d}{d \varepsilon}\left( g(x+ \varepsilon \delta x ,u+ \varepsilon \delta u) \right)
$$

$$
\begin{aligned}
\frac{d}{d \varepsilon}\left( g(x+ \varepsilon \delta x ,u+ \varepsilon \delta u) \right) &=
\left[ \frac{\partial g(x+ \varepsilon \delta x ,u+ \varepsilon \delta u)}{\partial (x+ \varepsilon \delta x)} \right] \frac{d}{d\varepsilon}(x+ \varepsilon \delta x) + \left[ \frac{\partial g(x+ \varepsilon \delta x ,u+ \varepsilon \delta u)}{\partial (u+ \varepsilon \delta u)} \right] \frac{d}{d\varepsilon}(u+ \varepsilon \delta u) \\
&= \left[ \frac{\partial g(x+ \varepsilon \delta x ,u+ \varepsilon \delta u)}{\partial (x+ \varepsilon \delta x)} \right] \delta x + \left[ \frac{\partial g(x+ \varepsilon \delta x ,u+ \varepsilon \delta u)}{\partial (u+ \varepsilon \delta u)} \right] \delta u \\ &= g_x \delta x + g_u \delta u
\end{aligned}
$$

よって、以下のようになる。

$$
\frac{d F(\varepsilon)}{d \varepsilon} = \delta \mu^T g(x+ \varepsilon \delta x ,u+ \varepsilon \delta u) + (\mu+ \varepsilon \delta \mu )^T \left(g_x \delta x + g_u \delta u \right)
$$

$\varepsilon = 0$ とすることで変分が計算できる。$\delta \mu^T g(x ,u)$はスカラーであるため、$\delta \mu^T g(x ,u) = g(x ,u)^T \delta \mu$となる。

$$
\begin{aligned}
\delta (\mu^T g(x,u)) &= \delta \mu^T g(x ,u) + (\mu)^T \left(g_x \delta x + g_u \delta u \right) \\
&= g(x ,u)^T \delta \mu  +  (\mu)^T g_x \delta x + (\mu)^T g_u \delta u \\
&= g(x ,u)^T \delta \mu +  (g_x^T \mu)^T  \delta x + (g_u^T \mu)^T \delta u 
\end{aligned}
$$

#### 拡大評価関数 $\bar{J}$ の形

不等式制約を組み込んだ変分を追加し、$\delta$ で整理を行った$\bar{J}$を示す。

$$
\begin{aligned}
\delta \bar{J} =& ({\Phi_x} - \lambda(T))^T \delta x(T) \\
& \quad + \int_0^T \left( \delta\lambda ^T(f - \dot{x}) 
+ (L_x + f_x^T \lambda  + \dot {\lambda } +g_x^T \mu )^T\delta x
+ (L_u + f_u ^T \lambda + g_u^T \mu)^T \delta u + g^T \delta \mu
\right) dt
\end{aligned}
$$


$\bar{J}$が停留する条件はその変分ゼロとなることである。

以下の変分は自由に決めることができるため、それらの係数をゼロとなるようにすると、以下の4つの式が導出される。

|変分|条件式|名前|
|:---|:---|:----|
|$\delta \lambda$| $f - \dot{x} = 0$|状態方程式|
|$\delta x(T)$| $\lambda(T) = \Phi_x(x(T))$| 終端条件|
|$\delta x$| $\dot{\lambda} = - (L_x + f_x^T \lambda + g_x^T \mu) $|随伴方程式|
|$\delta u$| $L_u + f_u^T \lambda + g_u^T \mu$|停留条件|


そして、$\mu, \delta \mu$ はKKT条件を満たすように決める必要がある。

$$
\begin{array}{l}
g(x,u) \le 0 \\
\mu \ge 0 \\
\mu^T g(x,u)  = 0
\end{array}
$$

#### 不等式制約付きPMPの必要条件

よって、拡大評価関数 $\bar{J}$ の停留条件を満たすため、下記の方程式全てを[0,T]の区間で満たすように、状態$x$, 制御入力$u$, 等式の随伴変数$\lambda$, 不等式の随伴変数$\mu$ を決定する必要がある。

$$
\begin{aligned}
&f - \dot{x} = 0 \\
&\lambda(T) = \Phi_x(x(T)) \\
&\dot{\lambda} = - (L_x + f_x^T \lambda + g_x^T \mu) \\
&L_u + f_u^T \lambda + g_u^T \mu = 0\\
&g(x,u) \le 0 \\
&\mu \ge 0 \\
&\mu_i g_i(x,u)  = 0 , \quad i = 1, \cdots , m
\end{aligned}
$$


### C/GMRESにおける考え方

等式制約の場合、状態 $x$ の時系列を制御入力 $u$ を用いて時間を前進させて作り、終端で $\lambda[N]$ を計算し、今度は時間を後退させて $\lambda$ の時系列を求め、それらから停留条件 $F=0$ の列を作成した。

不等式制約を追加した場合で考えると、時系列を次のように計算する。

$$
\begin{array}{l}
x[k+1] = x[k] + h f(x[k], u_k(t_k)) , \quad k = 0, \cdots , N-1 \\
\quad \darr \\
\lambda[N] = \Phi_x(x[N]) \\
\quad \darr \\
\lambda[k] = \lambda[k+1] + h (L_x(x[k], u_k(t_k)) + f_x^T(x[k], u_k(t_k))) \lambda[k+1] + g_x^T(x[k], u_k(t_k)) \mu[k] , \quad k=N-1 , \cdots 1
\end{array}
$$

そして、各時刻の停留条件を満たす必要がある。

$$
H_u = L_u(x[k], u[k]) + f_u^T(x[k], u[k]) \lambda[k+1] + g_u^T(x[k], u[k]) \mu[k] = 0 , \quad k = 0, \cdots N-1
$$

そして同時に以下のKKT条件も満たす必要がある。

$$
g(x[k], u[k]) \le 0 \\
\mu[k] \ge 0 \\
\mu[k]^T g(x[k], u_k(t_k)) = 0
$$

ここで二つの不等式があり、これが問題を困難にさせている。

$$
g(x[k], u[k]) \le 0 \\
\mu[k] \ge 0
$$

C/GMRESは基本的に 

$$
F(z) = 0
$$

という等式を解きたいので、この不等式では扱いにくい。

そこで下記の方法不等式を計算できる形へ変換することを行う。

- C/GMRESでよく使われるダミー変数による等式化
- interior-point法
- active-set法
- complementarityを平滑化して等式化する方法


### ダミー変数による等式化

一つの不等式制約として、制御入力の上限を以下のように設定することを考える。

$$
g_0(x,u) = u^2 \le u_{max}^2 \rightarrow g_0(x,u) = u^2 - u_{max}^2 \le 0 
$$

ここで $g_0$ は不等式であるため、ダミー変数 $u_d$ を用いて次のように等式化することを行う。

$$
C(u,u_d) = u^2 + u_d^2 - u_{max}^2 = 0
$$

上式は右辺の0を用いて変換すると以下のようになる。$u_d^2 \ge 0$ より、$u$と$u_{max}$の不等式制約の関係がある。

$$
u^2 = u_{max}^2 - u_d^2 \le u_{max}^2 \rightarrow u^2 \le u_{max}^2
$$

このダミー変数 $u_d$ が不明な値であるため、停留条件 $F$ の中で計算を行うようにする。


#### Hamiltonianの構成

次のように$H$を構成する。不等式制約 $g(x,u)$を等式制約 $C(u, u_d)$ へ置き換えているため、係数も$\mu$ではなく、$\rho$としている。

$$
H=L(x,u) + \lambda^T f(x,u) + \rho C(u, u_d)
$$

これを$x, u, u_d, \lambda, \rho$で偏微分する。

$$
\begin{array}{l}
H_\lambda = f(x,u) \\
H_x = L_x(x,u) + f_x^T(x,u) \lambda\\
H_u = L_u(x,u) + f_u^T(x,u)\lambda + C_u^T(u, u_d)\rho =0  \\
H_{u_d} = C_{u_d}^T(u, u_d) \rho = 0\\
H_{\rho} = C_{u_d}(u, u_d) = 0\\
\end{array}
$$

今回の等式条件が$C(u,u_d)$と$u$に関する関数であるため、$H_x$には表れていないことに注意すること。

これより、PMPの必要条件は以下のようになる。

運動方程式

$$
\dot{x} = H_\lambda
$$

終端条件

$$
\lambda(T) = \Phi_x(x(T))
$$

随伴方程式

$$
\dot{\lambda} = -H_x
$$

停留条件

$$
\begin{aligned}
H_u = 0 \\
H_{u_d} = 0 \\
H_{\rho} = 0
\end{aligned}
$$

$H_\rho=C=0$ が停留条件として加わっている。

##### $u_d$の2重解に対する対応

$H_{u_d}$を具体的に計算すると、以下となる。

$$
H_{u_d} = \rho \frac{\partial C}{\partial u_d} = 2 \rho u_d = 0 \rightarrow \rho u_d = 0
$$

$C(u,u_d) = u^2 + u_d^2 - u_{max}^2 = 0 $から$u_d$を計算すると、

$$
u_d = \pm \sqrt{u_{max}^2 - u^2}
$$

と同一の$u$ に対して 正負2つの解が存在する。また、$H_{u_d} = 2 \rho u_d = 0$も、この二つの解を区別する条件にはならない。

そこで、ランニングコスト$L$に$-r_d u_d , r_d > 0$ の項を追加した$L_d$を設定する。

$$
L_d(x,u,u_d) = L(x,u) - r_d u_d
$$

$-r_d u_d$ に $u_d = +8 , \space -8$ の二つの解を代入して比較すると、

- $u_d = +8$ なら $-r_d u_d = - 8 r_d$
- $u_d = -8$ なら $-r_d u_d = + 8 r_d$

$\bar{J}$の最小化問題を解いているので、$-8 r_d < +8 r_d$ より、選択すべきは $u_d = +8$ となる。

$-r_d u_d$ を評価関数に設定することで、$u_d > 0$の解を目的関数側から優先させることが出来る。

そして$H$は

$$
H=L_d(x,u,u_d) + \lambda^T f(x,u) + \rho C(u, u_d)
$$

より$H_{u_d}$は以下となる。

$$
H_{u_d} = -r_d + 2 \rho u_d = 0 \rightarrow \rho u_d = \frac{r_d}{2} > 0
$$

これは、最適化問題として $-r_d u_d$ によって $u_d>0$ が優先され、その解について

$$
\rho = \frac{r_d}{2 u_d}
$$

という関係が停留条件から決まる。

#### ダミー入力を考慮したHamiltonianの構成

Hamiltonian のランニングコストを$L$から$L_d$に変更する。

$$
H=L_d(x,u,u_d) + \lambda^T f(x,u) + \rho C(u, u_d)
$$

これを$x, u, u_d, \lambda, \rho$ で偏微分する。

$$
\begin{array}{l}
H_\lambda = f(x,u) \\
H_x = L_{d,x}(x,u, u_d) + f_x^T(x,u)\lambda  \\
H_u = L_{d,u}(x,u, u_d) + f_u^T(x,u)\lambda  + C_u^T(u, u_d)\rho = 0\\
H_{u_d} = L_{d,{u_d}}(x, u , u_d) + C_{u_d}^T(u, u_d) \rho = -r_d + 2 \rho u_d = 0\\
H_\rho = C(u, u_d) = 0
\end{array}
$$

これより、PMPの必要条件は以下のようになるが、上記の $-r_d u_d$ の補正がない式と同様であることに注目する。これがHamiltonianによる記述が効果的である理由である。

運動方程式

$$
\dot{x} = H_\lambda
$$

終端条件

$$
\lambda(T) = \Phi_x(x(T))
$$

随伴方程式

$$
\dot{\lambda} = -H_x
$$

停留条件

$$
\begin{aligned}
H_u = 0 \\ 
H_{u_d} = 0 \\
H_\rho = 0
\end{aligned}
$$

これにより、KKT条件の不等式制約 $g(x,u) \le 0$ をダミー入力を用いた等式制約 $C(u, u_d)=0$ へ変換し、PMPの停留条件と等式制約 $C=0$ を連立して扱える。
この連立方程式を停留条件 $F=0$ として構成することで、C/GMRESによる計算が可能となる。

注意点として、$-r_d u_d$ を追加することで、

$$
\min J (x,u) \quad \text{s.t} \quad |u| \le u_{max}
$$

と完全に同一の最適化問題ではなく、近似的に扱うものである。よって、$r_d$はできるだけ小さくすることが必要である。

#### C/GMRESでの計算

予測ホライゾンの離散時間を$\tau_0 , \cdots , \tau_{N-1}$ とすると、

実時間 $t_k$ における制御入力$U(t_k)$は以下であった。

$$
U(t_k) = [u_0(t_k), \cdots , u_{N-1}(t_k)]^T
$$

これと同様にダミー変数$u_d$も各時刻で次のような列になる。

$$
U_d(t_k) = [u_{d_0}(t_k), \cdots , u_{d_{N-1}}(t_k)]^T
$$

また、$\rho$も時系列となる。

$$
\rho(t_k) = [\rho_0(t_k), \cdots , \rho_{N-1}(t_k)]^T
$$

これらを用いて、停留条件は以下のように計算する。

状態$x$の時系列($x[0]$を設定して時間を進展させる。)

$$
x[n+1] = x[n] + h f(x[n], u_n(t_k)), \quad n = 0 , \cdots , N-1
$$

終端条件

$$
\lambda[N] = \Phi_x(x[N])
$$

随伴係数$\lambda$の時系列

$$
\lambda[n] = \lambda[n+1] + h \big[ L_{d,x}(x[n], u_n(t_k), u_{d_n}(t_k)) + f_x^T(x[n], u_n(t_k)) \lambda[n+1] \big] , \quad n=N-1 , \cdots 1
$$

そして、停留条件 $F=0$を次のように計算する。

$$
F = \begin{bmatrix}
H_{u,0} \\
H_{u_d,0}\\
H_{\rho, 0} \\
\vdots\\
H_{u,N-1} \\
H_{u_d,N-1}\\
H_{\rho,N-1} \\
\end{bmatrix} =
\begin{bmatrix}
L_{d,u}(x[0], u_0(t_k), u_{d_0}(t_k)) +  f_u^T(x[0], u[0]) \lambda[1] + C_u^T(u_0(t_k), u_{d_0}(t_k)) \rho_0(t_k) \\
L_{d,{u_d}}(x[0], u_0(t_k), u_{d_0}(t_k)) + C_{u_d}^T(u_0(t_k), u_{d_0}(t_k)) \rho_0(t_k) \\
C(u_0(t_k), u_{d_0}(t_k)) \\
\vdots \\
L_{d,u}(x[N-1], u_{N-1}(t_k), u_{d_{N-1}}(t_k)) + f_u^T(x[N-1], u[N-1]) \lambda[N]  + C_u^T(u_{N-1}(t_k), u_{d_{N-1}}(t_k)) \rho_{N-1}(t_k) \\
L_{d,{u_d}}(x[N-1], u_{N-1}(t_k), u_{d_{N-1}}(t_k)) + C_{u_d}^T(u_{N-1}(t_k), u_{d_{N-1}}(t_k)) \rho_{N-1}(t_k) \\
C(u_{N-1}(t_k), u_{d_{N-1}}(t_k)) \\
\end{bmatrix} = 0
$$

#### 停留条件の追跡

C/GMRESでは $F=0$の停留条件を時間が変化したとしても追従したいという制御である。そのため$d F /dt = 0$ という条件がある。

$F$は上記の計算する処理から、制御入力列 $U(t_k)$、ダミー入力列 $U_d(t_k)$、等式制約$C$の係数列 $\rho(t_k)$ と、状態の時系列の初期値 $x[0]=x(t_k)$から計算される。そして予測ホライゾンの時刻$T = T_f(1 - \exp(-\alpha t_k))$から求める 時間ステップ $h = T/N$ がある。

ここで $U(t_k)$を改めて以下のように設定する。

$$
U(t_k) = [u_0(t_k), u_{d_0}(t_k), \rho_0(t_k) , \cdots , u_{N-1}(t_k), u_{d_{N-1}}(t_k), \rho_{N-1}(t_k) ]^T
$$

これより、停留条件$F$は次のように表現される。

$$
F(U(t_k), x(t_k), t_k) = 0
$$

$d F / d t = 0 $ では数値誤差などでズレが生じるため、安定化を考慮したC/GMRESの計算は次のような式である。これは $F = F(0) \exp(-\zeta t)$ となるため、時間とともに"0"に収束していく。

$$
\frac{d F}{d t} = - \zeta F
$$

$F$の時間微分は$t_k=t$と捉えて、全微分を行うと、

$$
\begin{aligned}
\frac{d F}{d t} &= \frac{\partial F}{\partial U}\frac{d U}{dt} + \frac{\partial F}{\partial x}\frac{d x}{dt} + \frac{\partial F}{\partial t}\frac{d t}{dt}\\ 
&= F_u \dot{U} + F_x \dot{x} + F_t
\end{aligned}
$$

よって、制約$C$の無いのC/GMRES場合と同じく下記の線形方程式からGMRESで $\dot{U}$ を求めて、

$$
F_u \dot{U} = - F_x \dot{x} - F_t - \zeta F
$$

次の時刻の制御ループの制御入力列 $U(t_{k+1})$ を求める計算を行う。

$$
U(t_{k+1}) = U(t_k) + \Delta t \ \dot{U}(t_k)
$$

用いる有限差分はそれぞれ次にようになり、これも制約 $C$ がない場合と同じである。

GMRESの右辺$b_k$

$$
b_k = - \frac{F(U(t_k), x(t_k) + \varepsilon \dot{x}(t_k), t_k + \varepsilon ) - F(U(t_k), x(t_k), t_k)}{\varepsilon} - \zeta F(U(t_k), x(t_k), t_k) 
$$

GMRES計算の初期残差$r_0 = b_k - F_U \dot{U}^{(0)}(t_k)$に関しては

$$
F_U \dot{U}^{(0)}(t_k) = \frac{F(U(t_k)+ \varepsilon \dot{U}^{(0)}(t_k), x(t_k), t_k ) - F(U(t_k), x(t_k), t_k)}{\varepsilon}
$$

Arnoldi法の step m で用いる $F_U(t_k)v_m$は

$$
F_U(t_k)v_m = \frac{F(U(t_k)+ \varepsilon v_m ,x(t_k), t_k ) - F(U(t_k), x(t_k), t_k)}{\varepsilon}
$$

となる。よって、$U(t_k)$の構成と$F$の計算内容に変更があることが以外、制約$C$がない場合と同一の処理で行うことが出来る。


#### $U(t_k)$の初期値 $U(0)$を求める

方法は解析的に、または数値的に解く方法になる。

予測ホライゾンは以下のように、時間が進むにつれて伸びるようになっている。

$$
T(t_k) = T_f(1 - \exp(-\alpha t_k))
$$

このため、制御開始の $t_0 = 0$ では $T(0) = 0$ となる。

よって、予測ホライゾンの時刻 "0" につぶれているため、状態の時系列は以下となり、

$$
x_i(0) = x(0)
$$

そして、随伴変数も次のようになる。

$$
\lambda_i(0) = \Phi_x(x(0))
$$

そして、制御入力、ダミー変数、$\rho$もそれぞれ以下のように置く。

$$
u_i(0) = u(0), \quad u_{d,i}(0) = u_d(0), \quad \rho_i(0) = \rho(0)
$$

全てが 時刻 "0" の一点になっているため、以下の方程式を同時に満たすように $u(0), u_d(0), \rho(0)$ を求める。

$$
\begin{aligned}
H_u(x(0), \lambda(0), u(0), u_d(0), \rho(0)) = 0\\
H_{u_d}(x(0), u(0), u_d(0), \rho(0)) = 0 \\
C(u(0), u_d(0)) = 0
\end{aligned}
$$

##### 数値的、かつ解析的に求める方法

今回の制約は、

$$
C(u, u_d) = u^2 + u_d^2 - u_{max}^2 = 0
$$

そして、

$$
H_{u_d} = -r_d + 2 \rho u_d = 0
$$

である。よって、$u(0)$から$u_d(0)$は次のようになる。$-r_d , u_d$ の項より正の$u_d(0)$を選択している。

$$
u_d(0) = \sqrt{u_{max}^2 - u(0)^2}
$$

そしてこの $u_d(0)$ より $\rho(0)$ は次のように求まる。

$$
\rho(0) = \frac{r_d}{2 u_d(0)}= \frac{r_d}{2 \sqrt{u_{max}^2 - u(0)^2}}
$$

$H_u$は$L_d$の$u$による偏微分は$u_d$の項が消えることを考慮し、以下のようになる。

$$
H_u(x(0), \lambda(0), u(0), u_d(0), \rho(0)) = L_u(x(0), u(0)) + f_u^T(x(0), u(0))\Phi_x(x(0)) + 2 \rho(0) u(0) = 0
$$

ここの $\rho(0)$ を代入すると、未知数が $u(0)$だけの式になる。

$$
L_u(x(0), u(0)) + f_u^T(x(0), u(0))\Phi_x(x(0)) + \frac{r_d u(0)}{\sqrt{u_{max}^2 - u(0)^2}} = 0
$$

これをNewton-Raphson法などで解き、$u(0)$を求め、$u_d(0)$, $\rho(0)$ を計算する。

$$
\begin{array}{c}
u_d(0) = \sqrt{u_{max}^2 - u(0)^2} \\
\darr \\
\rho(0) = \dfrac{r_d}{2 u_d(0)}
\end{array}
$$

##### 数値的に解く

求めたい以下の3つ変数のNewton-Raphson法で解く。

$$
U(0) = [u(0), u_d(0), \rho(0)]^T
$$ 

多変数のNewton-Raphon法は下記の連立方程式を解く問題である。

$$
F(U) = 0
$$

$U$周りでTaylor展開を行うと、次のようになる。

$$
F(U + \Delta U) \approx F(U) + J_F(U)\Delta U
$$

現在の$F(U^{(j)}) \neq 0$ であるとして、ここから $U^{(j+1)} = U + \Delta U$ と動いた先が

$$
F(U^{(j+1)}) = 0
$$

となるようにしたいと考える。

するとTaylor展開を行った少し先の値が"0"となると考えるため、

$$
F(U^{(j)}) + J_F(U^{(j)})\Delta U = 0
$$

以下の式を$\Delta U$について解き、$U^{(j+1)} = U^{(j)} + \Delta U$ と更新して残差 $||F(U^{(j+1)})|| < \sigma$ を確認する。 

$$
J_F(U^{(j)})\Delta U = - F(U^{(j)})
$$

$J_F$は以下のヤコビアンであるため、LU分解などで解くことが出来る。

$$
J_F = \frac{\partial F}{\partial U} =
\begin{bmatrix}
\dfrac{\partial H_u}{\partial u} & \dfrac{\partial H_u}{\partial u_d} & \dfrac{\partial H_u}{\partial \rho} \\ 
\dfrac{\partial H_{u_d}}{\partial u} & \dfrac{\partial H_{u_d}}{\partial u_d} & \dfrac{\partial H_{u_d}}{\partial \rho} \\ 
\dfrac{\partial C}{\partial u} & \dfrac{\partial C}{\partial u_d} & \dfrac{\partial C}{\partial \rho} \\ 
\end{bmatrix}
$$
