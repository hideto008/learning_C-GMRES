# 不等式制約の考慮

## KKT 条件の再考

KKT条件は不等式制約 $g(x) \le 0$ のもと、関数 $f(x)$ の最小値を探す問題である。

その必要条件として以下の四つの式を満たす必要がある。[KKT](KKT.ipynb)のNoteでは随伴変数を$\lambda$で記載していたが、KKT条件をPMPへ組み込むことを考慮し、$\mu$としている。

||式|名前|
|:---|:---|:---|
|(1)|$\nabla f + \mu \nabla g = 0$| Stationarity (停留条件)|
|(2)|$g \le 0$|Primal feasibility (主問題の実行可能性)|
|(3)|$\mu \ge  0$|Dual feasibility (双対実行可能性)|
|(4)|$\mu g = 0$| Complementary slackness(相補性条件)|

重要なのは四つ目の相補性条件 $\mu g =0 $ であり、これは、制約条件の境界上であれば $g=0$ となり、境界上でなければ相補性条件を満たすため$\mu = 0$ となるものである。

もう少し説明を追加する。 $x$の状態により、$g = -2$ のように制約の内側にあるときは $x$ は自由に動くことができる。これを制約が非アクティブであるという。ここで $\mu g = 0$ の相補性条件より、$\mu=0$ でなければならない。
また、$g=0$ と制約の境界上にいる場合、制約の外側、つまり $g > 0$ へ動こうとすると、制約の内側へ戻そうとする力$\mu \ge 0$ により$g=0$ へと押し留まるような意味になる。これをアクティブであるという。このとき相補性条件 とDual feasibility より、$\mu \ge 0$ となる。

## PMPにおける最適制御問題の再考

不等式制約 $g(x,u) \le 0$ を組み込んだ最適制御問題は下記のようになる。

$$
\begin{array}{c}
\min\limits_{u} \ J = \Phi(x(T)) + \int_0^T L(x,u) dt \\
\text{subject to } \quad \begin{split} &\dot{x} = f(x,u) \\ &g(x,u) \le 0 \end{split} 
\end{array}
$$

不等式制約 $g(x,u)$は下記のように複数の条件であり、それに伴い$\mu$もベクトルとなる。

$$
g(x,u) = \begin{bmatrix}
g_1(x,u) \\ \vdots \\ g_m(x,u) 
\end{bmatrix} \in \mathbb{R}^{m} , \quad 
\mu = \begin{bmatrix}
\mu_1 \\ \vdots \\ \mu_m
\end{bmatrix} \in \mathbb{R}^m
$$

[PMP](PMP.ipynb)のときと同様に、拡大評価関数$\bar{J}$を構成すると以下のようになる。

$$
\bar{J} = \Phi(x(T)) + \int_0^T \left[ L(x,u) + \lambda^T(f(x,u) - \dot{x}) + \mu^T g(x,u)\right] dt
$$

$\bar{J}$の変分を計算する。

まず、状態 $x$, 制御入力 $u$, ラグランジュ未定乗数 $\lambda$ , $\mu$ をそれぞれ以下のように微小変化させる。

$$
\begin{aligned}
x &\rightarrow x + \varepsilon \delta x \\
\dot{x} &\rightarrow \dot{x} + \varepsilon \delta \dot{x} \\
x(T) &\rightarrow x(T) + \varepsilon \delta x(T) \\
u &\rightarrow u + \varepsilon \delta u \\
\lambda &\rightarrow \lambda + \varepsilon \delta \lambda \\
\mu &\rightarrow \mu + \varepsilon \delta \mu \\
\end{aligned}
$$

したがって、微小変化を代入した拡大評価関数 $\bar{J}$ は以下となる。

$$
\begin{aligned}
\bar{J} & \left[x + \varepsilon \delta x, u + \varepsilon \delta u, \lambda + \varepsilon \delta \lambda,  \mu + \varepsilon \delta \mu  \right] \\
&= \Phi(x(T)+ \varepsilon \delta x(T))  \\ 
 &+  \int_0^T \bigg[ L(x+ \varepsilon \delta x ,u+ \varepsilon \delta u) \\
 &\quad + (\lambda+ \varepsilon \delta \lambda )^T \bigg(f(x+ \varepsilon \delta x ,u+ \varepsilon \delta u) - \dot{x}- \varepsilon \delta \dot{x} \bigg)  \\
 &\qquad + (\mu+ \varepsilon \delta \mu )^T g(x+ \varepsilon \delta x ,u+ \varepsilon \delta u )\bigg] dt
\end{aligned}
$$


変分$\delta$を計算するとは、以下のように、$\varepsilon$ で微分した後、$\varepsilon=0$ と代入する計算であった。

$$
\delta = \left. \frac{d}{d \varepsilon} \right|_{\varepsilon=0}
$$

微小変化を代入した$\bar{J}$の変分を行う。

左辺は以下となる。

$$
\delta \bar{J} = \left. \frac{d}{d \varepsilon} \bar{J}[x + \varepsilon \delta x, u + \varepsilon \delta u, \lambda + \varepsilon \delta \lambda,  \mu + \varepsilon \delta \mu ] \right|_{\varepsilon=0}
$$

右辺は以下となる。

$$
\begin{aligned}
\delta \bar{J} &= \dfrac{d}{d \varepsilon} \Bigg[ \Phi(x(T)+ \varepsilon \delta x(T)) + \int_0^T \bigg[ L(x+ \varepsilon \delta x ,u+ \varepsilon \delta u) \\
 & \quad + (\lambda+ \varepsilon \delta \lambda )^T \bigg(f(x+ \varepsilon \delta x ,u+ \varepsilon \delta u) - \dot{x}- \varepsilon \delta \dot{x} \bigg)  \\
 & \qquad + (\mu+ \varepsilon \delta \mu )^T g(x+ \varepsilon \delta x ,u+ \varepsilon \delta u) \left. \bigg] dt \Bigg] \right|_{\varepsilon=0}
\end{aligned} 
$$

上式は等式制約でのPMPを求める際に、 $\mu^T g(x,u)$の項以外をすでに求めている。よって、以下の変分を考える。

$$
\left. \frac{d}{d \varepsilon} \left[ (\mu+ \varepsilon \delta \mu )^T g(x+ \varepsilon \delta x ,u+ \varepsilon \delta u \right]  \right|_{\varepsilon=0}
$$

等式制約に対するラグランジュ乗数$\lambda$は自由に動かすこと(変分)ができるが、不等式制約に対する $\mu$ は $\mu \ge 0$ という Dual feasability の制約があり、自由に変分することが出来ない。よって、KKT条件を考慮する必要がある。

## 不等式制約を組み込むときの注意

上記では不等式制約 $g(x,u) \le 0$ を 拡張評価関数 $\bar{J}$ へ $\mu^T g(x,u)$ として組み込んだ。

これは $g(x,u)$ がスカラーではなくベクトルであり、不等式制約が複数設定してあるということになる。

$x = (x_1, x_2)$ と状態が複数ある場合、不等式制約は通常 $g_i(x,u) = x_1^2 + x_2^2 -u^2 - 1 \le 0$ のようなスカラーで評価する形をとる。

よって、KKT条件で上げた4式はスカラーの表現である。これを複数の不等式制約とする場合、4式がそれぞれの制約で組み込まれる。

$$
\begin{split}
\begin{array}{l}
\nabla f(x,u) +J_g(x,u)^T \mu = 0 \\
g_i (x,u) \le 0 \\
\mu_i g_i(x,u) = 0 \\
\mu_i \ge 0
\end{array}
\end{split} \quad, i = 1, \cdots, m
$$

式中の$J_g(x,u)^T$は $\nabla g_i$ で構成されたヤコビアンであり、

$$
J_g(x,u) =
\begin{bmatrix}
\nabla g_1 ^T \\
\vdots \\
\nabla g_m ^T
\end{bmatrix} =
\begin{bmatrix}
\dfrac{\partial g_1}{\partial x} & \dfrac{\partial g_1}{\partial u} \\
\vdots & \vdots\\
\dfrac{\partial g_m}{\partial x} & \dfrac{\partial g_m}{\partial u} \\
\end{bmatrix}
$$

$\mu$は$g_i(x,u)$のそれぞれと積算されるスカラー$\mu_i$で構成されたベクトルである。

$$
\mu = \begin{bmatrix} \mu_1 & \cdots & \mu_m\end{bmatrix}^T
$$

よって $J_g(x,u)^T {\mu}$ は以下のようになる。

$$
J_g(x,u)^T {\mu}=
\begin{bmatrix}
\dfrac{\partial g_1}{\partial x} & \cdots & \dfrac{\partial g_m}{\partial x} \\
\dfrac{\partial g_1}{\partial u} & \cdots & \dfrac{\partial g_m}{\partial u} \\
\end{bmatrix}
\begin{bmatrix}
\mu_1 \\ \vdots \\ \mu_m 
\end{bmatrix} = 
\begin{bmatrix}
\mu_1 \dfrac{\partial g_1}{\partial x} + \cdots + \mu_m \dfrac{\partial g_m}{\partial x} \\
\vdots \\
\mu_1 \dfrac{\partial g_1}{\partial u} + \cdots + \mu_m \dfrac{\partial g_m}{\partial u}
\end{bmatrix} =
\mu_1 \begin{bmatrix}
\dfrac{\partial g_1}{\partial x} \\ \dfrac{\partial g_1}{\partial u}
\end{bmatrix} + \cdots +
\mu_m \begin{bmatrix}
\dfrac{\partial g_m}{\partial x} \\ \dfrac{\partial g_m}{\partial u}
\end{bmatrix} 
$$


### 不等式制約に関する変分

$$
\delta ( \mu^T g) = \left. \frac{d}{d \varepsilon} \left[ (\mu+ \varepsilon \delta \mu )^T g(x+ \varepsilon \delta x ,u+ \varepsilon \delta u \right]  \right|_{\varepsilon=0}
$$

上式の計算を積の微分とチェインルールにより計算する。$F(\varepsilon)$ を以下のようにする。$g(x,u), \mu$はそれぞれベクトルであることに注意する。

$$
F(\varepsilon) = (\mu+ \varepsilon \delta \mu )^T g(x+ \varepsilon \delta x ,u+ \varepsilon \delta u)
$$

$$
\frac{d F(\varepsilon)}{d \varepsilon} = \delta \mu^T g(x+ \varepsilon \delta x ,u+ \varepsilon \delta u) + (\mu+ \varepsilon \delta \mu )^T \frac{d}{d \varepsilon}\left( g(x+ \varepsilon \delta x ,u+ \varepsilon \delta u) \right)
$$

$$
\begin{aligned}
\frac{d}{d \varepsilon}\left( g(x+ \varepsilon \delta x ,u+ \varepsilon \delta u) \right) &=
\left[ \frac{\partial g(x+ \varepsilon \delta x ,u+ \varepsilon \delta u)}{\partial (x+ \varepsilon \delta x)} \right] \frac{d}{d\varepsilon}(x+ \varepsilon \delta x) + \left[ \frac{\partial g(x+ \varepsilon \delta x ,u+ \varepsilon \delta u)}{\partial (u+ \varepsilon \delta u)} \right] \frac{d}{d\varepsilon}(u+ \varepsilon \delta u) \\
&= \left[ \frac{\partial g(x+ \varepsilon \delta x ,u+ \varepsilon \delta u)}{\partial (x+ \varepsilon \delta x)} \right] \delta x + \left[ \frac{\partial g(x+ \varepsilon \delta x ,u+ \varepsilon \delta u)}{\partial (u+ \varepsilon \delta u)} \right] \delta u \\ &= g_x \delta x + g_u \delta u
\end{aligned}
$$

よって、以下のようになる。

$$
\frac{d F(\varepsilon)}{d \varepsilon} = \delta \mu^T g(x+ \varepsilon \delta x ,u+ \varepsilon \delta u) + (\mu+ \varepsilon \delta \mu )^T \left(g_x \delta x + g_u \delta u \right)
$$

$\varepsilon = 0$ とすることで変分が計算できる。$\delta \mu^T g(x ,u)$はスカラーであるため、$\delta \mu^T g(x ,u) = g(x ,u)^T \delta \mu$となる。

$$
\begin{aligned}
\delta (\mu^T g(x,u)) &= \delta \mu^T g(x ,u) + (\mu)^T \left(g_x \delta x + g_u \delta u \right) \\
&= g(x ,u)^T \delta \mu  +  (\mu)^T g_x \delta x + (\mu)^T g_u \delta u \\
&= g(x ,u)^T \delta \mu +  (g_x^T \mu)^T  \delta x + (g_u^T \mu)^T \delta u 
\end{aligned}
$$

#### 拡大評価関数 $\bar{J}$ の形

不等式制約を組み込んだ変分を追加し、$\delta$ で整理を行った$\bar{J}$を示す。

$$
\begin{aligned}
\delta \bar{J} =& ({\Phi_x} - \lambda(T))^T \delta x(T) \\
& \quad + \int_0^T \left( \delta\lambda ^T(f - \dot{x}) 
+ (L_x + f_x^T \lambda  + \dot {\lambda } +g_x^T \mu )^T\delta x
+ (L_u + f_u ^T \lambda + g_u^T \mu)^T \delta u + g^T \delta \mu
\right) dt
\end{aligned}
$$


$\bar{J}$が停留する条件はその変分ゼロとなることである。

以下の変分は自由に決めることができるため、それらの係数をゼロとなるようにすると、以下の4つの式が導出される。

|変分|条件式|名前|
|:---|:---|:----|
|$\delta \lambda$| $f - \dot{x} = 0$|状態方程式|
|$\delta x(T)$| $\lambda(T) = \Phi_x(x(T))$| 終端条件|
|$\delta x$| $\dot{\lambda} = - (L_x + f_x^T \lambda + g_x^T \mu) $|随伴方程式|
|$\delta u$| $L_u + f_u^T \lambda + g_u^T \mu$|停留条件|


そして、$\mu, \delta \mu$ はKKT条件を満たすように決める必要がある。

$$
\begin{array}{l}
g(x,u) \le 0 \\
\mu \ge 0 \\
\mu^T g(x,u)  = 0
\end{array}
$$

よって、拡大評価関数 $\bar{J}$ の停留条件を満たすため、下記の方程式全てを[0,T]の区間で満たすように、状態$x$, 制御入力$u$, 等式の随伴変数$\lambda$, 不等式の随伴変数$\mu$ を決定する必要がある。

$$
\begin{aligned}
&f - \dot{x} = 0 \\
&\lambda(T) = \Phi_x(x(T)) \\
&\dot{\lambda} = - (L_x + f_x^T \lambda + g_x^T \mu) \\
&L_u + f_u^T \lambda + g_u^T \mu \\
&g(x,u) \le 0 \\
&\mu \ge 0 \\
&\mu^T g(x,u)  = 0
\end{aligned}
$$


### C/GMRESにおける考え方

等式制約の場合、状態 $x$ の時系列を制御入力 $u$ を用いて時間を前進させて作り、終端で $\lambda[N]$ を計算し、今度は時間を後退させて $\lambda$ の時系列を求め、それらから停留条件 $F=0$ の列を作成した。

不等式制約を追加した場合で考えると、時系列を次のように計算し、

$$
\begin{array}{l}
x[k+1] = x[k] + h f(x[k], u_k(t_k)) , \quad k = 0, \cdots , N-1 \\
\quad \darr \\
\lambda[N] = \Phi_x(x[N]) \\
\quad \darr \\
\lambda[k] = \lambda[k+1] + h (L_x(x[k], u_k(t_k)) + f_x^T(x[k], u_k(t_k))) \lambda[k+1] + g_x^T(x[k], u_k(t_k)) \mu[k] , \quad k=N-1 , \cdots 1
\end{array}
$$

そして、各時刻の停留条件を満たす必要がある。

$$
H_u = L_u(x[k], u[k]) + f_u^T(x[k], u[k]) \lambda[k+1] + g_u^T(x[k], u[k]) \mu[k] = 0 , \quad k = 0, \cdots N-1
$$

そして同時に以下のKKT条件も満たす必要がある。

$$
g(x[k], u[k]) \le 0 \\
\mu[k] \ge 0 \\
\mu[k]^T g(x[k], u_k(t_k)) = 0
$$

ここで二つの不等式があり、これが問題を困難にさせている。

$$
g \le 0 \\
\mu \ge 0
$$

C/GMRESは基本的に 

$$
F(z) = 0
$$

という等式を解きたいので、この不等式では扱いにくい。

そこで下記の方法不等式を計算できる形へ変換することを行う。

- active-set法
- interior-point法
- complementarityを平滑化して等式化する方法
- C/GMRESでよく使われるダミー変数による等式化
