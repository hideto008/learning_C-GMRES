# オイラー法

C/GMRESでは状態 $x(t)$ と随伴変数 $\lambda(t)$ の時系列を数値計算で求めていく。<br>
そこで使用されるのがオイラー法である。

このNoteでは、オイラー法の説明と、具体的な計算の例を示す。

## 理論


一階常微分方程式の解を数値計算により離散的に求めることを考える。

### 前進オイラー法

$$
\dot{y}(t) = F(y(t), t)
$$

上式を左辺と右辺を離散化すると以下となる。

$$
\frac{y[k+1] - y[k]}{h} = F(y[k], t[k])
$$

ここから、現在時刻 $k$ から一つ次の時刻の$k+1$ の値は次のように求めることが出来る。

$$
y[k+1] = y[k] + h \ F(y[k], t[k])
$$

これを前進オイラー法と呼ぶ。

### 時間的な後退

前進オイラー法を時間方向に進めることは ”次の値 = 現在値 + $\Delta t \space \times$  現在の微分値" という関係になる。<br>
これを時間を逆方向に進めることを考える。

$$
y[k] = y[k+1] + \Delta t \ \dot{y}(t_{k+1})
$$ 

現在値は$y(t_{k+1})$ であり、この微分値は以下となる。

$$
\dot{y}(t_{k+1}) = F(y(t_{k+1}), t_{k+1})
$$

時間的な後退は刻み幅を $-h$ と考えるので、現在時刻 $k+1$ から一つ前の時刻 $k$ の値は次のように求めることが出来る。

$$
y[k] = y[k+1] - h \ F(y[k+1], t[k+1])
$$

これを時間を逆方向に進めるオイラー法と呼ぶ。

## 例題：質点モデルの時系列

以下の質点モデルの運動方程式を考える。

$$
f(t) = m \frac{ d^2 x(t)}{d t^2}
$$

$\dfrac{d x(t)}{dt} = v(t)$ であるため、上式は以下のようになる。

$$
f(t) = m \frac{dv(t)}{dt}
$$

一階常微分方程式の形に書き換える。

$$
\begin{aligned}
\frac{d x(t)}{dt} &= v(t) \\
\frac{dv(t)}{dt} &= \frac{1}{m} f(t) \\
\end{aligned}
$$

状態を$[x(t), v(t)]^T$ とすると、上式は以下となる。

$$
\frac{d}{dt}
\begin{bmatrix}
x(t) \\ v(t)
\end{bmatrix} =
\begin{bmatrix}
0 & 1 \\
0 & 0 \\
\end{bmatrix}
\begin{bmatrix}
x(t) \\ v(t)
\end{bmatrix} +
\begin{bmatrix}
0 \\ 1/m
\end{bmatrix}f(t)
$$

### 前進オイラー法による時系列の計算

上式を離散化すると次のようになる。

$$
\begin{aligned}
\begin{bmatrix}
x[k+1] \\ v[k+1]
\end{bmatrix} &=
\begin{bmatrix}
x[k] \\ v[k]
\end{bmatrix} +
h \left(
\begin{bmatrix}
0 & 1 \\
0 & 0 \\
\end{bmatrix}
\begin{bmatrix}
x[k] \\ v[k]
\end{bmatrix} +
\begin{bmatrix}
0 \\ 1/m
\end{bmatrix}f[k]
\right) \\
&=\begin{bmatrix}
1 & h \\
0 & 1 \\
\end{bmatrix}
\begin{bmatrix}
x[k] \\ v[k]
\end{bmatrix} +
\begin{bmatrix}
0 \\ h/m
\end{bmatrix}f[k]
\end{aligned}
$$

この方法により、$[x[0], v[0]]^T$から、$[x[N], v[N]]^T$までの時系列を求めることが出来る。

### 時間を逆方向に進めるオイラー法による時系列の計算

$$
\begin{aligned}
\begin{bmatrix}
x[k] \\ v[k]
\end{bmatrix} &=
\begin{bmatrix}
x[k+1] \\ v[k+1]
\end{bmatrix} -
h \left(
\begin{bmatrix}
0 & 1 \\
0 & 0 \\
\end{bmatrix}
\begin{bmatrix}
x[k+1] \\ v[k+1]
\end{bmatrix} +
\begin{bmatrix}
0 \\ 1/m
\end{bmatrix}f[k+1]
\right) \\
&=
\begin{bmatrix}
1 & -h \\
0 & 1 \\
\end{bmatrix}
\begin{bmatrix}
x[k+1] \\ v[k+1]
\end{bmatrix} +
\begin{bmatrix}
0 \\ -h/m
\end{bmatrix}f[k+1]
\end{aligned}
$$

この方法により、$[x[N], v[N]]^T$から、$[x[0], v[0]]^T$までの時系列を求めることが出来る。


## C/GMRESへの繋がり

$$
x[0] \rightarrow x[1] \rightarrow \cdots \rightarrow x[N]
